# Rede Direcionada

```python
class Node:
  def __init__(self, id):
    self.id = id
    self.neighbors = []

  def add_neighbor(self, neighbor_id):
    self.neighbors.append(neighbor_id)

  def degree(self):
    return len(self.neighdors)

  def __repr__(self):  # For unambiguous representation
    return f"Node(id={self.id}, neighbors={self.neighbors})"


# Node(id=1, neighbors=[[2, 7], [3, 4]])
```

```python
class DirectedNetwork:
  def __init__(self):
    self.nodes = {}

  def add_node(self, node):
    self.nodes[node.id] = node

  def add_link(self, node1_id, node2_id):
    if node1_id not in self.nodes:
      self.add_node(Node(node1_id))
    if node2_id not in self.nodes:
      self.add_node(Node(node2_id))
    self.nodes[node1_id].add_neighbor(node2_id)


  def get_node(self,node):
    if node in self.nodes:
      return self.nodes[node]
    return None

  # SE FOREM REDES SEPARADAS(DIRECIONADA E NÃO DIRECIONADA):

  def get_density(self):
    quant_nodes = len(self.nodes)

    potential_links = quant_nodes*(quant_nodes-1)

    actual_links = 0
    for node in self.nodes.values():
      neighbors = node.neighbors
      actual_links += len(neighbors)
    density = actual_links/potential_links
    return(density)
```

```python
network = DirectedNetwork()

network.add_node(Node(1))
network.add_node(Node(2))
network.add_node(Node(3))
network.add_node(Node(4))
network.add_node(Node(5))

network.add_link(1,2)
network.add_link(1,3)
network.add_link(3,4)
network.add_link(5,4)
network.get_density()

```
